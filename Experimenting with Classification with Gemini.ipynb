```python
from pathlib import Path

def image_format(image_path):
    img = Path(image_path)

    if not img.exists():
        raise FileNotFoundError(f"Could not find image: {img}")

    image_parts = [
        {
            "mime_type": "image/jpeg", ## Mime type are PNG - image/png. JPEG - image/jpeg. WEBP - image/webp
            "data": img.read_bytes()
        }
    ]
    return image_parts

def gemini_output(image_path, system_prompt, user_prompt):

    image_info = image_format(image_path)
    input_prompt= [system_prompt, image_info[0], user_prompt]
    response = model1.generate_content(input_prompt)
    return response.text
```

```python
import os
from PIL import Image
from IPython.display import display, Image as IPImage

class Quiz:
    def __init__(self, base_dir):
        self.base_dir = base_dir
        self.user_query = None
        self.user_answers = []

    def get_subtopic(self):
        self.user_query = input("Which subtopic do you want questions from? ")

    def get_number_of_files(self, directory):
        return len(os.listdir(directory))

    def run_quiz(self):
        if not self.user_query:
            self.get_subtopic()

        direc = os.path.join(self.base_dir, self.user_query)
        number_files = self.get_number_of_files(direc)

        for x in range(1, number_files + 1):
            img_path = os.path.join(self.base_dir, self.user_query, f'Q{x}-{self.user_query}.png')

            # Check if image file exists to avoid potential errors
            if os.path.exists(img_path):
                im = Image.open(img_path)
                display(IPImage(img_path))  # Display the image
            else:
                print(f"Image {img_path} not found.")

            # Take user input and store in the vector
            user_ans = input("Answer: ")
            self.user_answers.append(user_ans)
            print(" ")

        self.display_user_answers()

    def display_user_answers(self):
        print("User Answers:", self.user_answers)

base_dir = '/content/drive/MyDrive/Alevel Physics'
quiz = Quiz(base_dir)
quiz.run_quiz()
```

```python
import os
from PIL import Image
from IPython.display import display, Image as IPImage

# Get the list of files in the directory
lst = os.listdir('/content/drive/MyDrive/Alevel Physics/EST')  # your directory path
number_files = len(lst)
print(number_files)

# Get user input for subtopic
user_query = input("Which subtopic do you want questions from? ")
anslist = []
iscorrect = []

# Construct directory path
direc = "/content/drive/MyDrive/Alevel Physics/{}".format(user_query)
lst = os.listdir(direc)
number_files = len(lst)

# Initialize variables
x = 1
user_answers = []

while x <= number_files:
    img_path = '/content/drive/MyDrive/Alevel Physics/{}/Q{}-{}.png'.format(user_query, x, user_query)

    # Check if image file exists to avoid potential errors
    if os.path.exists(img_path):
        with Image.open(img_path) as im:
            fixed_width = 600
            fixed_height = 150
            im_resized = im.resize((fixed_width, fixed_height), Image.LANCZOS)
            im_resized.save('/tmp/resized_image.png')
            display(IPImage(filename='/tmp/resized_image.png'))
    else:
        print(f"Image {img_path} not found.")

    # Take user input and store in the vector
    user_ans = input("Answer: ")
    user_answers.append(user_ans)

    print(" ")
    system_prompt = """
                    You are an expert Alevel Examiner who will check the student's answer. If the student's answer is correct then just say Its Correct. If its incorrect then explain why is it incorrect and what should be the correct answer?
                    """
    image_path = img_path
    user_prompt = "Explain"
    gemini_output(image_path, system_prompt, user_prompt)
    x += 1

# Now user_answers contains all the user inputs
print("User Answers:", user_answers)
```
