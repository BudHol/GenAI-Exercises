```python
%%writefile app.py
from dotenv import load_dotenv
import streamlit as st
import google.generativeai as genai
import os

os.environ['GOOGLE_API_KEY']=""
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
genai.configure(api_key=GOOGLE_API_KEY)
model = genai.GenerativeModel("gemini-pro")
load_dotenv()

genai.configure(api_key="")

# Set up the model
generation_config = {
    "temperature": 0.9,
    "top_p": 1,
    "top_k": 1,
    "max_output_tokens": 2048,
}

safety_settings = [
    {
        "category": "HARM_CATEGORY_HARASSMENT",
        "threshold": "BLOCK_MEDIUM_AND_ABOVE"
    },
    {
        "category": "HARM_CATEGORY_HATE_SPEECH",
        "threshold": "BLOCK_MEDIUM_AND_ABOVE"
    },
    {
        "category": "HARM_CATEGORY_SEXUALLY_EXPLICIT",
        "threshold": "BLOCK_MEDIUM_AND_ABOVE"
    },
    {
        "category": "HARM_CATEGORY_DANGEROUS_CONTENT",
        "threshold": "BLOCK_MEDIUM_AND_ABOVE"
    },
]

model = genai.GenerativeModel(model_name="gemini-1.0-pro-001",
                              generation_config=generation_config,
                              safety_settings=safety_settings)
prompt_parts = [
    "You are a Pseudocode Question Generator for Alevel Computer Science students. Teachers will use you by giving you a pseudocode concept or multiple concepts. You will then make CAIE International Alevel Computer Science caliber questions (you may refer to past papers for example questions) .",
    "Concept(s): Functions and Procedures",
    "Question When a video is selected, the program gives an estimate of the file size of the converted video.\nThe estimate in kilobytes is calculated by multiplying:\nthe number of pixels in the video's resolution by...\nthe number of frames per second by...\nthe length of the video in minutes by...\nthe value 0.0013.\n(c) Write a function in pseudocode that estimates the size of a converted video.\nIt should:\ntake in 3 parameters: pixels, framesPerSec, lengthMins\ncalculate the estimated file size\nreturn a string with the file size, including units\nuse megabytes for sizes under 1000 megabytes, otherwise the estimate should be given\nin gigabytes.\nExamples:\n480000 pixels at 24 frames per second for 60 minutes will return a size of 898.56 MB\n480000 pixels at 24 frames per second for 120 minutes will return a size of 1.797 GB.",
    "Concept(s): 1D and 2D Arrays",
    "Question A shop sells tiles of the following dimensions (in centimetres):\n- 10 x 10\n- 15 x 15\n- 20 x 20\nThe manager wants to calculate the cost of tiling the shop floor.\nThe dimensions of the shop floor are: 600cm by 400cm.\n(d) It is decided before tiling that the tiles will be laid on their sides, so that the lengths of the tiles will be horizontal. Write pseudocode to create a 2D array to store the tiles on the floor.  The array should represent the dimensions of the shop floor.\n(e) Write pseudocode that will calculate the number of tiles needed to cover the floor. You should assume that the 10 x 10 tiles will be used to cover the remaining area.\n(f) Each 10 x 10 tile costs 20p, each 15 x 15 tile costs 35p and each 20 x 20 tile costs 50p. Write pseudocode to calculate the cost of the tiles used to cover the floor.",
]

def generate_content(concepts_input):
    prompt_parts[1] = f"Concept(s): {concepts_input}"
    response = model.generate_content(prompt_parts)
    return response.text

def main():
    st.title("Pseudocode Question Generator")
    st.write("Enter concept(s) (or 'exit' to stop):")

    concepts_input = st.text_input("Concept(s):")

    if st.button("Generate Question"):
        if concepts_input.lower() == 'exit':
            st.warning("Exiting...")
        else:
            result = generate_content(concepts_input)
            st.subheader("Generated Question:")
            st.code(result, language="plaintext")

if __name__ == "__main__":
    main()
```
